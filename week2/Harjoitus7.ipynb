```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
#a)
t = np.linspace(0, 10, 1000)
omega = 2  # rad/s
x = np.cos(omega * t)
y = np.sin(omega * t)
f = 3 * np.sin(2 * t +1)
#b)
f1 = 3 * np.cos(2 * t +1 - np.pi/2)
f2 = 3 * np.sin(7 * t - 1)

#c)
A = 3/2         #amplitudi
omega = 7       #kulmataajuus
phi = -1         #aloitusvaihe
T = 2 * np.pi / omega  #jaksonaika

#d)
A1 = 3
T1 = 4
phi1 = np.pi
omega1 = 2*np.pi/T1 #=pi/2
#f(t) = 3sin(pi/2t+pi)


plt.figure(figsize=(15,8))
plt.plot(t, x, label="x(t) = cos(2t)")
plt.plot(t, y, label="y(t) = sin(2t)")

plt.plot(t, f, label="f(t)=3sin(2t+1)")
plt.plot(t, f1, label="f(t)=3cos(2t+1-π/2)") #siirretään 90 astetta

plt.xlabel("Aika t (s)")
plt.ylabel("Koordinaatti")
plt.title("Pisteen liike yksikköympyrällä")

plt.grid(True)
plt.legend()
plt.show()
```
