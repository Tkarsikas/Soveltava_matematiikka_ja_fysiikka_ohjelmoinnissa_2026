```python
#Ladataan kirjastot, jos et ole vielä niitä asentanut, asenna pipillä komentorivill
#pip install numpy pandas matplotlib
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt

#Jos haluaa muuttaa värejä:
from cycler import cycler
import colorcet as cc
plt.rcParams['axes.prop_cycle'] = cycler(color=cc.glasbey)

#Luetaan data Pandas DataFrameen
df = pd.read_csv('./Data/Raw Data.csv')
#df['Time (s)'] # Valitsee kyseisen sarakkeen Time (s)
df

#Piirretään kuvaaja Matplotlib.pyplot -kirjastolla
plt.figure(figsize=(15,8))
plt.plot(df['Time (s)'],df['Linear Acceleration x (m/s^2)'],'k' )
plt.plot(df['Time (s)'],df['Linear Acceleration y (m/s^2)'],'g')
plt.plot(df['Time (s)'],df['Linear Acceleration z (m/s^2)'],'b')
plt.plot(df['Time (s)'],df['Absolute acceleration (m/s^2)'],'y')
plt.title('Kiihtyvyysmittaus')
plt.legend(['x','y','z'])
plt.grid()
plt.ylabel('Kiihtyvyys [m/$s^2$]')
plt.xlabel('Aika [s]')
plt.axis([0,11,-17,17])
plt.show()
#plt.plot(df['Linear Acceleration x (m/s^2)'])
#df

#Koordinaattiakselit:
# y: ylös-alas
# x: oikealle - vasemmalle
# z: eteen - taakse
```
