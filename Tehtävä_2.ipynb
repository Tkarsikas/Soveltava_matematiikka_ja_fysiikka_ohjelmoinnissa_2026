 e) Laske komponenttien ja kokonaiskiihtyvyyden keskiarvot ja keskihajonnat.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('./Data/Raw Data.csv')
df
plt.figure(figsize=(15,8))
plt.plot(df['Time (s)'], df['Linear Acceleration x (m/s^2)'],'k')
plt.plot(df['Time (s)'],df['Linear Acceleration y (m/s^2)'])
plt.plot(df['Time (s)'],df['Linear Acceleration z (m/s^2)'])
plt.plot(df['Time (s)'],df['Absolute acceleration (m/s^2)'])
plt.title('Kiihtyvyysmittaus')
plt.legend(['x','y','z','abs'])
plt.grid()
plt.ylabel("Kiihtyvyys [m/$s^2$]")
plt.xlabel("aika [s]")
plt.axis([0,8,-21,21])
plt.show()
```
