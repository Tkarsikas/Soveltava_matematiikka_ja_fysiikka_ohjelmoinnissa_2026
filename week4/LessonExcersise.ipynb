```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#genedoiraan testidata
t = np.arange(0,30,0.1) #aika välillä 0-30 s, sämplätty 0.1s (10hz)
taajuus1 = 0.5
taajuus2 = 1
f = np.sin(taajuus1*2*np.pi*t) + np.cos(taajuus2*2*np.pi*t) + np.random.randn(len(t))*2

plt.figure(figsize=(12,4))
plt.plot(t,f)
plt.title("Testisignaali")
```

```python
#Lasketaan Fourier-muunnos:
#Aika t
#Signaali f
N = len(f) #datapisteiden lukumäärä signaalissa
dt = t[1]-t[0] #Näytteenottoväli, datapisteiden ajallinen välimatka

#Lasketaan Fourier
fourier = np.fft.fft(f,N)

#Lasketaan tehospektri (Power spectral density, PSD)
psd = fourier*np.conj(fourier)/N

freq = 1/(dt*N)*np.arange(N) #Taajuudet
L = np.arange(1, np.floor(N/2), dtype='int') #Määritellään ne tehospektrin arvot, joita käytetään


PSD = np.array([freq[L], psd[L].real]) #Taajuudet ja tehospektrin arvot

plt.plot(PSD[0,:], PSD[1,:])
plt.grid()
plt.title('Tehospektri')
plt.xlabel('Taajuus')
plt.ylabel('Teho')
```
